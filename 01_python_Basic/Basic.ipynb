```python
# determine if year is a leap year or not
year = int(input("Enter a year:"))
if(year%4 ==0):
    if(year%100 == 0):
        if(year%400 == 0):
            print("The year is a leap year")
        else:
            print("The year is not a leap year")
    else:
        print("The year is a leap year")
else:
    print(year,"The year is not a leap year")
```

```python
# determine if a number is prime or not
num = int(input("Enter a number:"))
is_prime = True

for i in range(2, num):
    if num % i == 0:
        is_prime = False
        break

if is_prime:
    print(f"{num} is a prime number.")
else:
    print(f"{num} is not a prime number.")
```

```python
# nested loops
for i in range(3):
    for j in range(2):
        print(f"i: {i}, j: {j}")
```

```python
# nested loops to print a pattern
rows = 5
for i in range(1, rows + 1):
    for j in range(1, i + 1):
        print("*", end="")
    print()
```

```python
# nested loops to print a pattern
rows = 5
columns = 5
for i in range(1, rows + 1):
    for j in range(1, columns + 1):
        print("*", end="")
    print()
```

```python
list_fruit = ["apple", "banana", "orange", "grape", "mango"]
print(list_fruit[0])  # Accessing the first element
print(list_fruit[1])  # Accessing the second element
print(list_fruit[0:3])  # Accessing a slice of the list 
print(list_fruit[-1])  # Accessing the last element
print(list_fruit[1:4])  # Accessing a slice of the list
print(list_fruit[1:4:2])  # Accessing a slice of the list with a step of 2
```

```python
# slicing
list1 = [1, 2, 3, 4, 5]
print(list1[1:4])  # Printing elements from index 1 to 3
print(list1[:3])   # Printing elements from the beginning to index 2
print(list1[2:])   # Printing elements from index 2 to the end
print(list1[::2])  # Printing every second element
print(list1[::-1])  # Printing the list in reverse order
print(list1[1:4:2])  # Printing elements from index 1 to 3 with a step of 2
print(list1[::])  # Printing the entire list
```

```python
# iterating through a list
for number in list1:
    print(number)
```

```python
# iterating with index
for index, number in enumerate(list1):
    print(f"Index: {index}, Number: {number}")
```
